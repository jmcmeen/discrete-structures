### Problem 1: Truth Values

Use Python to evaluate and print the truth value of each proposition: $3 \ge 5$, &nbsp;'9 is odd', &nbsp;and $7 < 4$.

```python
print("3 >= 5  :", 3 >= 5)
print("9 is odd:", 9 % 2 == 1)
print("7 < 4   :", 7 < 4)
```

### Problem 2: Negation

Given `p = True` and `q = False`, print the negations $\lnot p$ and $\lnot q$.

```python
p = True
q = False

print("not p:", not p)
print("not q:", not q)
```

### Problem 3: Truth Table for AND and OR

Build a truth table for $p \land q$ and $p \lor q$ over all combinations of $p$ and $q$. (Hint: `itertools.product([True, False], repeat=2)`.)

```python
from itertools import product

print(f"{'p':>6} {'q':>6} {'p and q':>8} {'p or q':>7}")
for p, q in product([True, False], repeat=2):
    print(f"{str(p):>6} {str(q):>6} {str(p and q):>8} {str(p or q):>7}")
```

### Problem 4: Implication Truth Table

In Python the conditional $p \rightarrow q$ is `(not p) or q`. Print its full truth table over all values of $p$ and $q$.

```python
from itertools import product

print(f"{'p':>6} {'q':>6} {'p -> q':>7}")
for p, q in product([True, False], repeat=2):
    print(f"{str(p):>6} {str(q):>6} {str((not p) or q):>7}")
```
