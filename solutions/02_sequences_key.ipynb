### Problem 1: Floor and Ceiling

Use Python's `math` module to print the floor $\lfloor x \rfloor$ and ceiling $\lceil x \rceil$ of each value in `values`.

```python
import math
values = [3.8, -2.5, 4.0, math.pi]

for x in values:
    print(f"x={x:6.3f}   floor={math.floor(x):>3}   ceil={math.ceil(x):>3}")
```

### Problem 2: Terms of a Sequence

The sequence is given explicitly by $a_n = 3n + 2$. Print the first ten terms $a_1, a_2, \dots, a_{10}$.

```python
def a(n):
    return 3 * n + 2

for n in range(1, 11):
    print(f"a_{n} = {a(n)}")
```

### Problem 3: A Recurrence Relation

An account starts with \$2000 and earns 6% interest compounded annually, so $a_n = 1.06\,a_{n-1}$ with $a_0 = 2000$. Compute the balance after 50 years.

```python
balance = 2000.0
rate = 0.06

for year in range(1, 51):
    balance = balance * (1 + rate)

print(f"Balance after 50 years: ${balance:,.2f}")
```

### Problem 4: Summations

Use Python to evaluate each summation: $\sum_{k=1}^{4}(2k-1)$, $\sum_{n=0}^{2}(3n+2)$, and $\sum_{m=1}^{3} m^2$.

```python
print("sum_{k=1}^{4} (2k-1) =", sum(2*k - 1 for k in range(1, 5)))
print("sum_{n=0}^{2} (3n+2) =", sum(3*n + 2 for n in range(0, 3)))
print("sum_{m=1}^{3} m^2    =", sum(m**2 for m in range(1, 4)))
```
