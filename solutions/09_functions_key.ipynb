### Problem 1: Range of a Function

The function $f(n)$ assigns to each nonnegative integer its last digit ($n \bmod 10$). Compute the **range** of $f$ over the domain $\{0, 1, \dots, 40\}$.

```python
def f(n):
    return n % 10

range_of_f = sorted({f(n) for n in range(41)})
print("range:", range_of_f)
```

### Problem 2: Is it One-to-One?

Determine whether each function is one-to-one (injective) on the domain $\{-5, \dots, 5\}$: $f(x) = 3x + 2$ and $g(x) = x^2$. A function is one-to-one exactly when no two inputs share an output.

```python
def is_one_to_one(func, domain):
    outputs = [func(x) for x in domain]
    return len(set(outputs)) == len(outputs)

domain = range(-5, 6)
print("f(x) = 3x + 2 one-to-one:", is_one_to_one(lambda x: 3 * x + 2, domain))  # True
print("g(x) = x^2   one-to-one:", is_one_to_one(lambda x: x ** 2, domain))       # False
```

### Problem 3: The Floor Function

Compute $f(x) = \lfloor x^2 / 3 \rfloor$ for each $x$ in $\{-3, -2, -1, 0, 1, 2, 3\}$.

```python
import math
xs = [-3, -2, -1, 0, 1, 2, 3]
print([math.floor(x ** 2 / 3) for x in xs])
```

### Problem 4: Bytes to Store Bits

How many bytes are required to store $n$ bits of data, where one byte is 8 bits? Compute it for $n = 9, 20, 700, 9000$. (Hint: $\lceil n/8 \rceil$.)

```python
import math
for n in [9, 20, 700, 9000]:
    print(f"{n:>5} bits -> {math.ceil(n / 8)} bytes")
```
