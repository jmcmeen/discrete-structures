### Problem 1: Modular Arithmetic

Use Python's `%` operator to compute each value: $12 \bmod 5$, $57 \bmod 3$, $-30 \bmod 13$, and $-17 \bmod 18$.

```python
for a, m in [(12, 5), (57, 3), (-30, 13), (-17, 18)]:
    print(f"{a} mod {m} = {a % m}")
```

### Problem 2: GCD and LCM

Using `math.gcd` and `math.lcm`, compute $\gcd(864, 800)$ and $\operatorname{lcm}(864, 800)$, then verify that $\gcd \times \operatorname{lcm} = 864 \times 800$.

```python
import math
a, b = 864, 800

g, l = math.gcd(a, b), math.lcm(a, b)
print("gcd =", g)
print("lcm =", l)
print("gcd * lcm =", g * l, " and  a * b =", a * b, " equal?", g * l == a * b)
```

### Problem 3: Base Conversion

Use Python's built-ins to (a) convert the binary string `'10010110'` to base 10, (b) show 150 in binary, and (c) show 239 in hexadecimal.

### Problem 4: Caesar Cipher

Write a function that shifts each letter of a message by a key (a Caesar cipher). Encrypt `\"Discrete Math\"` with a shift of 3, then decrypt the result back.

```python
def caesar(message, key):
    result = ""
    for ch in message:
        if ch.isalpha():
            base = ord("A") if ch.isupper() else ord("a")
            result += chr((ord(ch) - base + key) % 26 + base)
        else:
            result += ch
    return result

encrypted = caesar("Discrete Math", 3)
print("Encrypted:", encrypted)
print("Decrypted:", caesar(encrypted, -3))
```
