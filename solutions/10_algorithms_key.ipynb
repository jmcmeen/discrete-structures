### Problem 1: Linear Search

Write a function `linear_search(lst, target)` that returns the index of `target` in `lst`, or $-1$ if it is not present. Test it on the list in the starter.

```python
data = [4, 8, 15, 16, 23, 42]

def linear_search(lst, target):
    for i, x in enumerate(lst):
        if x == target:
            return i
    return -1

print("index of 15:", linear_search(data, 15))
print("index of 99:", linear_search(data, 99))
```

### Problem 2: Binary Search

Write `binary_search(lst, target)` for a **sorted** list, returning the index of `target` or $-1$. It should halve the search range each step.

```python
data = [1, 3, 5, 7, 9, 11, 13]

def binary_search(lst, target):
    lo, hi = 0, len(lst) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if lst[mid] == target:
            return mid
        elif lst[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

print("index of 9:", binary_search(data, 9))
print("index of 8:", binary_search(data, 8))
```

### Problem 3: Bubble Sort

Implement `bubble_sort` to return a new sorted copy of a list by repeatedly swapping adjacent out-of-order elements.

```python
data = [64, 34, 25, 12, 22, 11, 90]

def bubble_sort(a):
    a = a[:]
    n = len(a)
    for i in range(n):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

print(bubble_sort(data))
```

### Problem 4: Find the Maximum

Write an algorithm `find_max(a)` that returns the largest element of a non-empty list by scanning it once. What is its big-O time complexity?

```python
data = [3, 41, 17, 99, 8]

def find_max(a):
    largest = a[0]
    for x in a[1:]:
        if x > largest:
            largest = x
    return largest

print("max:", find_max(data))
# It examines each element once, so it runs in O(n) time.
```
