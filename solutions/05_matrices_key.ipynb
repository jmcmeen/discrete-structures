### Problem 1: Matrix Addition

Using NumPy, compute $A + B$ for the matrices defined in the starter.

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A + B)
```

### Problem 2: Matrix Multiplication

Compute the matrix product $A B$ (use the `@` operator). Is $AB$ equal to $BA$?

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A @ B =\n", A @ B)
print("B @ A =\n", B @ A)
print("AB == BA ?", np.array_equal(A @ B, B @ A))   # False: matrix multiplication is not commutative
```

### Problem 3: Transpose

Print the transpose $A^{T}$ of the matrix $A$.

```python
import numpy as np
A = np.array([[1, 2, 3], [4, 5, 6]])

print(A.T)
```

### Problem 4: Scalar Multiplication and the Identity

Compute $3A$, and print the $3 \times 3$ identity matrix.

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])

print("3A =\n", 3 * A)
print("3x3 identity =\n", np.eye(3, dtype=int))
```
