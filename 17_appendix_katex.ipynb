# Appendix: Math Notation Reference (KaTeX)

These notebooks render mathematics with [KaTeX](https://katex.org/), a fast LaTeX-style math renderer. Write math **inline** with single dollar signs — `$A \cup B$` renders as $A \cup B$ — and as a **centered display** equation with double dollar signs:

`$$\sum_{i=1}^{n} i = \frac{n(n+1)}{2}$$`

$$\sum_{i=1}^{n} i = \frac{n(n+1)}{2}$$

Use this appendix as a quick lookup for the notation used throughout the course. In every table, **Type this** is the literal LaTeX you write and **Renders as** is the result.

A matrix uses `bmatrix` (square brackets) or `pmatrix` (parentheses), with `&` between columns and `\\` between rows:

`\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}`

$$\begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$$

Multi-line derivations use `aligned`, lining up at the `&`:

$$\begin{aligned} (a+b)^2 &= a^2 + 2ab + b^2 \\ &= a^2 + b^2 + 2ab \end{aligned}$$
